# Assignment 3: Multimodal AI Model Exploration

This notebook demonstrates multiple input-output modalities using different AI providers. Set API keys in a local `.env` file or as Colab secrets before running the cells.

## Model Plan

| Modality | Provider | Model |
|---|---|---|
| Text -> Text | Groq | `llama-3.3-70b-versatile` |
| Text -> Image | Hugging Face | `stabilityai/stable-diffusion-xl-base-1.0` |
| Image -> Text | Google Gemini | `gemini-1.5-flash` |
| Text -> Audio | Hugging Face | `facebook/mms-tts-eng` |
| Audio -> Text | Groq | `whisper-large-v3-turbo` |
| Text -> Video | Hugging Face | `damo-vilab/text-to-video-ms-1.7b` |
| Video -> Text | Google Gemini | `gemini-1.5-flash` |

```python
# If running in a fresh notebook environment, uncomment this line:
# !pip install -r requirements.txt

from pathlib import Path
from IPython.display import Audio, Image, Video, display, Markdown

from multimodal_models import (
    describe_image,
    describe_video,
    generate_audio,
    generate_image,
    generate_text_response,
    generate_video,
    transcribe_audio,
)
```

## 1. Text -> Text

Sample input: a student asks for a simple explanation of multimodal AI.

```python
text_prompt = "Explain multimodal AI in simple language with one real-world example."
text_output = generate_text_response(text_prompt)
display(Markdown(text_output))
```

## 2. Text -> Image

Sample input: a prompt describing a futuristic classroom.

```python
image_prompt = "A futuristic classroom where students test multimodal AI tools, bright digital art"
image_result = generate_image(image_prompt)
Path(image_result.filename).write_bytes(image_result.data)
display(Image(data=image_result.data))
```

## 3. Image -> Text

Upload or provide a local image path. The generated image from the previous section can be reused.

```python
image_path = Path(image_result.filename)
image_description = describe_image(
    image_path.read_bytes(),
    "image/png",
    prompt="Describe this image and identify the important objects."
)
display(Markdown(image_description))
```

## 4. Text -> Audio

Sample input: a short sentence converted to speech.

```python
speech_prompt = "This is a generated audio sample for Assignment 3."
audio_result = generate_audio(speech_prompt)
Path(audio_result.filename).write_bytes(audio_result.data)
display(Audio(data=audio_result.data, autoplay=False))
```

## 5. Audio -> Text

The audio generated above is transcribed back into text.

```python
transcript = transcribe_audio(audio_result.filename)
display(Markdown(transcript))
```

## 6. Text -> Video

This section may take several minutes depending on the Hugging Face endpoint.

```python
video_prompt = "A short cinematic clip of a robot assistant arranging books in a modern library"
video_result = generate_video(video_prompt)
Path(video_result.filename).write_bytes(video_result.data)
display(Video(data=video_result.data, embed=True))
```

## 7. Video -> Text

The generated video is summarized using a multimodal Gemini model.

```python
video_summary = describe_video(
    Path(video_result.filename).read_bytes(),
    "video/mp4",
    prompt="Summarize this video in 5 bullet points."
)
display(Markdown(video_summary))
```
